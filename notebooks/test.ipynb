```python
import sys

sys.path.append("../pyds")
```

```python
import pyds
```

```python
settings = pyds.Settings()
```

```python
feat = pyds.Featurize(settings=settings)
```

```python
data = pyds.Data(settings=settings)
```

```python
response = data.upload_data(
    file_path="./sample_molecular_data.csv",
    filename="sample_molecular_data.csv",
    description="Sample Molecular Data"
)
```

```python
response
```

```python
dataset_addr = response["dataset_address"]
```

```python
result = feat.run(
    dataset_address=dataset_addr,
    featurizer="ecfp",
    output="test_featurized_output1",
    dataset_column="smiles",
    feat_kwargs={"radius": 2, "size": 1024},
    label_column="target",
)
```
